```python
def csv_to_pandas(filepath):
    df = pd.read_csv(filepath)
    return df
```

```python
# load and convert to pandas
charging_sessions = csv_to_pandas("./Data Share/charging_sessions.csv")
weather_burbank_airport = csv_to_pandas("./Data Share/weather_burbank_airport.csv")
charging_sessions.info()
```

```python
# expand userinput data
def parse_list(cell):
    if pd.isna(cell):
        return [] 
    return ast.literal_eval(cell)

charging_sessions["parsed"] = charging_sessions["userInputs"].apply(parse_list)

charging_sessions["last"] = charging_sessions["parsed"].apply(
    lambda lst: lst[-1] if len(lst) > 0 else None
)

expanded_df = pd.json_normalize(charging_sessions["last"])
expanded_df = expanded_df.drop(columns= "userID")
charging_sessions = pd.concat([charging_sessions, expanded_df], axis=1)

```

```python
# convert to local time, the timezone for all the values is America/Los_Angeles

time_cols = [
    'connectionTime', 
    'disconnectTime', 
    'doneChargingTime', 
    'modifiedAt', 
    'requestedDeparture'
]

for col in time_cols:
    charging_sessions[col] = (
        pd.to_datetime(charging_sessions[col], utc=True, errors='coerce') 
        .dt.tz_convert('America/Los_Angeles')                              
        .dt.tz_localize(None)                                              
    )
```

```python
# remove duplicate rows and useless columns
charging_sessions = charging_sessions[[
        "id",
        "connectionTime",
        "disconnectTime",
        "doneChargingTime",
        "kWhDelivered",
        "sessionID",
        "siteID",
        "stationID",
        "userID",
        "WhPerMile",
        "kWhRequested",
        "milesRequested",
        "minutesAvailable",
        "modifiedAt",
        "requestedDeparture"
    ]]
charging_sessions= charging_sessions.drop_duplicates(keep='first')
```

```python
# add missing values in temperature data by interpolation
numeric_cols = ['temperature', 'felt_temperature', 'windspeed', 'pressure']
weather_burbank_airport[numeric_cols] = weather_burbank_airport[numeric_cols].interpolate(method='linear')
```

```python
# only get the data which overlaps for weather and charging
weather_burbank_airport['timestamp'] = pd.to_datetime(weather_burbank_airport['timestamp'])

cs_min = charging_sessions['connectionTime'].min()
cs_max = charging_sessions['connectionTime'].max()

weather_min = weather_burbank_airport['timestamp'].min()
weather_max = weather_burbank_airport['timestamp'].max()

overlap_start = max(cs_min, weather_min)
overlap_end = min(cs_max, weather_max)

charging_sessions_overlap = charging_sessions[
    (charging_sessions['connectionTime'] >= overlap_start) &
    (charging_sessions['connectionTime'] <= overlap_end)
]

weather_burbank_airport_overlap = weather_burbank_airport[
    (weather_burbank_airport['timestamp'] >= overlap_start) &
    (weather_burbank_airport['timestamp'] <= overlap_end)
]
```

```python
# KPI 1 Hourly KWH delivered by Site
KPI1_df = charging_sessions.copy()

KPI1_df['hour'] = KPI1_df['connectionTime'].dt.floor('h')

hourly_kwh = (
    KPI1_df
    .groupby(['siteID', 'hour'], as_index=False)['kWhDelivered']
    .sum()
    .rename(columns={'kWhDelivered': 'kWh_hourly'})
)

site_ids = hourly_kwh['siteID'].unique()
fig, axes = plt.subplots(len(site_ids), 1, figsize=(15, 5 * len(site_ids)), sharex=True)


for ax, site_id in zip(axes, site_ids):
    site_data = hourly_kwh.query('siteID == @site_id')
    
    ax.bar(site_data['hour'], site_data['kWh_hourly'], width=0.03, color='skyblue', edgecolor='black', alpha=0.8)
    
    ax.set_title(f'Hourly kWh Delivered – Site {site_id}')
    ax.set_xlabel('Time')
    ax.set_ylabel('kWh Delivered')
    
    tick_interval = max(len(site_data)//12, 1)
    ax.set_xticks(site_data['hour'].iloc[::tick_interval])
    ax.set_xticklabels([h.strftime('%Y-%m-%d') for h in site_data['hour'].iloc[::tick_interval]], rotation=45, ha='right')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Clean Setup
df = charging_sessions.copy()
df['connectionTime'] = pd.to_datetime(df['connectionTime'])

# 2. Define "Valid Intent" (The User's "Order")
# We only care about users who ask for physically possible amounts (10-100 kWh)
valid_intent = df[
    (df['kWhRequested'] >= 10) & 
    (df['kWhRequested'] <= 100)
].copy()

# 3. Create the Weekly Timeline
# We align every timestamp to the start of its week
valid_intent['week_start'] = valid_intent['connectionTime'].dt.to_period('W').dt.start_time

# 4. Filter for "Scale" (Site Activity)
# Calculate total requested energy per site, per week
weekly_volume = valid_intent.groupby(['siteID', 'week_start'])['kWhRequested'].sum().reset_index()

# Identification: Which site-weeks have enough volume to be statistically significant? (> 26.5 kWh)
active_weeks = weekly_volume[weekly_volume['kWhRequested'] >= 26.5]

# Filter: Keep only the sessions that belong to these "Active Weeks"
# (This acts as an inner join without the complexity of merging timestamps)
final_df = valid_intent.merge(
    active_weeks[['siteID', 'week_start']], 
    on=['siteID', 'week_start'], 
    how='inner'
)

# 5. Calculate the KPI
# Sum up the energy for the remaining valid sessions
kpi_df = final_df.groupby(['siteID', 'week_start'])[['kWhDelivered', 'kWhRequested']].sum().reset_index()

# The Formula: Total Delivered / Total Requested
kpi_df['Reliability_Score'] = (kpi_df['kWhDelivered'] / kpi_df['kWhRequested']) * 100

# 6. Visualization
plt.figure(figsize=(15, 6))
sns.lineplot(data=kpi_df, x='week_start', y='Reliability_Score', hue='siteID', marker='o')

plt.axhline(100, color='red', linestyle='--', label='Ideal Target (100%)')
plt.axhline(80, color='green', linestyle=':', label='Good Service (80%)') # Added a realistic benchmark

plt.title('Hub Reliability Score: Delivered vs. Valid Intent', fontsize=14)
plt.ylabel('Fulfillment % (Weighted)', fontsize=12)
plt.xlabel('Week', fontsize=12)
plt.legend(title='Site ID', loc='upper right')
plt.grid(True, alpha=0.3)
plt.ylim(0, 120) # Clean up the y-axis
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# 1. Prepare the Data
proof_df = charging_sessions.copy()

# Ensure we have the parsed 'requestedDeparture' (from previous steps)
# If not, ensure you run the parsing code first! 
# We also need 'doneChargingTime' and 'connectionTime'.

# Filter 1: Only Registered Users (we need requestedDeparture)
proof_df = proof_df[proof_df['requestedDeparture'].notna()].copy()

# Filter 2: Only Valid Energy Requests
proof_df = proof_df[proof_df['kWhRequested'] > 0].copy()

# 2. Calculate Key Metrics

# A. Fulfillment Ratio
proof_df['fulfillment_ratio'] = (proof_df['kWhDelivered'] / proof_df['kWhRequested']) * 100
proof_df['fulfillment_ratio'] = proof_df['fulfillment_ratio'].clip(upper=120) # Cap for visibility

# B. Estimate "Time Needed" (Theoretical)
# We calculate the average charging speed for this specific session
# Speed = Delivered / (Done - Connect)
# Note: We use 'doneChargingTime' to capture the active period. 
proof_df['active_duration_h'] = (proof_df['doneChargingTime'] - proof_df['connectionTime']).dt.total_seconds() / 3600
proof_df['avg_power_kw'] = proof_df['kWhDelivered'] / proof_df['active_duration_h']

# Handle edge cases (divide by zero or near zero duration)
proof_df = proof_df[proof_df['active_duration_h'] > 0.1] 
proof_df = proof_df[proof_df['avg_power_kw'] < 100] # Remove outliers

# C. Calculate "Theoretical Time Needed" to meet the FULL request
proof_df['time_needed_for_full_request'] = proof_df['kWhRequested'] / proof_df['avg_power_kw']

# D. Calculate "Available Duration" (Stay Duration)
# How long did they actually stay? (Disconnect - Connect)
proof_df['stay_duration_h'] = (proof_df['disconnectTime'] - proof_df['connectionTime']).dt.total_seconds() / 3600

# E. SLACK TIME (The Magic Metric)
# Slack = (How long they stayed) - (How long they needed)
proof_df['slack_hours'] = proof_df['stay_duration_h'] - proof_df['time_needed_for_full_request']


# 3. Visualization: The "Constraint Proof" Plot
plt.figure(figsize=(12, 7))

# We use a scatter plot with transparency
plt.scatter(
    proof_df['slack_hours'], 
    proof_df['fulfillment_ratio'], 
    alpha=0.3, 
    c='purple', 
    edgecolors='none',
    s=20
)

plt.axvline(0, color='red', linestyle='--', label='Zero Slack (Time Limit)')
plt.axhline(100, color='green', linestyle='--', label='100% Fulfillment')

plt.title('Proof of Time Constraint: Slack Time vs. Fulfillment Ratio')
plt.xlabel('Slack Time (Hours)\n(Negative = Left Early | Positive = Had Extra Time)')
plt.ylabel('Fulfillment Ratio (%)')
plt.xlim(-5, 10) # Focus on the relevant range
plt.ylim(0, 120)
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# Group A: Successful Sessions (>= 98% Fulfillment)
success_group = proof_df[proof_df['fulfillment_ratio'] >= 98]
median_slack_success = success_group['slack_hours'].median()

# Group B: Failed Sessions (< 80% Fulfillment)
failed_group = proof_df[proof_df['fulfillment_ratio'] < 80]
median_slack_fail = failed_group['slack_hours'].median()

print(f"Median Slack for Successful Sessions: {median_slack_success:.2f} Hours")
print(f"Median Slack for Failed Sessions: {median_slack_fail:.2f} Hours")
```

```python
# KPI 3 percentage of spots with a car which is not charging

df = charging_sessions.copy()


df['doneChargingTime'] = df['doneChargingTime'].fillna(df['disconnectTime'])

total_stations = df.groupby('siteID')['stationID'].nunique().to_dict()

df['start_time'] = df['doneChargingTime']
df['end_time'] = df['disconnectTime']

df = df[df['end_time'] > df['start_time']].copy()

rows = []

for _, row in df.iterrows():
    session_start = row['start_time']
    session_end = row['end_time']
    site_id = row['siteID']
    
    start_hour_bucket = session_start.floor('h')
    end_hour_bucket = session_end.floor('h')
    
    hours_active = pd.date_range(start=start_hour_bucket, end=end_hour_bucket, freq='h')
    
    for hour in hours_active:
        hour_start = hour
        hour_end = hour + pd.Timedelta(hours=1)
        
        overlap_start = max(session_start, hour_start)
        overlap_end = min(session_end, hour_end)
        
        seconds = (overlap_end - overlap_start).total_seconds()
        
        if seconds > 0:
            fraction_hours = seconds / 3600.0
            rows.append({
                'siteID': site_id,
                'hour': hour,
                'fraction': fraction_hours
            })

usage_df = pd.DataFrame(rows)

hourly_usage = usage_df.groupby(['siteID', 'hour'])['fraction'].sum().reset_index()

hourly_usage['capacity_percent'] = hourly_usage.apply(
    lambda x: (x['fraction'] / total_stations.get(x['siteID'], 1)) * 100, axis=1
)

site_ids = hourly_usage['siteID'].unique()


fig, axes = plt.subplots(len(site_ids), 1, figsize=(15, 5 * len(site_ids)), sharex=False)

for ax, site_id in zip(axes, site_ids):

    site_data = hourly_usage[hourly_usage['siteID'] == site_id].sort_values('hour')
    
    ax.bar(site_data['hour'], site_data['capacity_percent'], 
           width=0.04, color='orange', edgecolor='black', alpha=0.7)
    
    ax.set_title(f'Hourly Blocking Capacity (%) – Site {site_id}')
    ax.set_ylabel('Blocked Capacity (%)')
    ax.set_ylim(0, 100)
    

    if len(site_data) > 0:

        tick_interval = max(len(site_data) // 12, 1)
        ax.set_xticks(site_data['hour'].iloc[::tick_interval])
        ax.set_xticklabels([h.strftime('%Y-%m-%d %H:%M') for h in site_data['hour'].iloc[::tick_interval]], 
                           rotation=45, ha='right')

plt.xlabel('Time')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import seaborn as sns

def cluster_and_plot(df, cluster_features, scatter_x, scatter_y, n_clusters=3):
    """
    Performs K-Means clustering and visualizes the result as a scatter plot.
    
    Parameters:
    - df: The DataFrame containing the data.
    - cluster_features: List of column names used for the math (clustering).
    - scatter_x: Column name for the X-axis of the plot.
    - scatter_y: Column name for the Y-axis of the plot.
    - n_clusters: Number of clusters (k).
    """
    
    # 1. Prepare Data
    data = df.copy()
    
    # Scale the features (Essential for K-Means)
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(data[cluster_features])
    
    # 2. Run K-Means
    kmeans = KMeans(n_clusters=n_clusters, random_state=42, n_init=10)
    data['cluster'] = kmeans.fit_predict(X_scaled)
    
    # 3. Visualize
    plt.figure(figsize=(10, 6))
    
    sns.scatterplot(
        data=data,
        x=scatter_x,
        y=scatter_y,
        hue='cluster',
        palette='viridis',
        alpha=0.7,
        s=100  # Dot size
    )
    
    plt.title(f'K-Means Clustering: {scatter_x} vs. {scatter_y}')
    plt.xlabel(scatter_x)
    plt.ylabel(scatter_y)
    plt.grid(True, alpha=0.3)
    plt.show()
    
    return data 
```

```python
def find_optimal_k(df, cluster_features):
    """
    Plots the 'Elbow Curve' to determine the optimal number of clusters.
    """
    # 1. Prepare Data
    data = df[cluster_features].copy()
    
    # Scale Data
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(data)
    
    # 2. Calculate Inertia for k=1 to k=10
    inertia = []
    K_range = range(1, 11)
    
    print("Calculating optimal k... this might take a moment.")
    
    for k in K_range:
        kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
        kmeans.fit(X_scaled)
        inertia.append(kmeans.inertia_)
        
    # 3. Plotting
    plt.figure(figsize=(10, 6))
    plt.plot(K_range, inertia, 'bo-', markersize=8)
    plt.xlabel('Number of Clusters (k)')
    plt.ylabel('Inertia (Sum of Squared Distances)')
    plt.title('The Elbow Method: Look for the "Bend"')
    plt.xticks(K_range)
    plt.grid(True)
    plt.show()


find_optimal_k(charging_sessions, ['start_sin', 'duration_hours', 'start_cos'])
```

```python
import plotly.express as px
def cluster_and_plot_3d(df, cluster_cols, k, x_col, y_col, z_col, symbol_col=None):
    """
    Auto-detects k, clusters data, and returns a 3D plot.
    
    Parameters:
    - df: Dataframe
    - cluster_cols: List of cols for the math (e.g. ['sin', 'cos', 'duration'])
    - x_col, y_col, z_col: Cols for the 3D visual axes
    - symbol_col: (Optional) Col for dot shapes
    """
    data = df.copy()
    
    # 1. Standardize the "Math" columns
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(data[cluster_cols])
    

    
    # 3. Apply Final Clustering
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    data['cluster'] = kmeans.fit_predict(X_scaled).astype(str)
    
    # 4. Visualize
    fig = px.scatter_3d(
        data,
        x=x_col,
        y=y_col,
        z=z_col,
        color='cluster',
        symbol=symbol_col,
        title=f'3D Clusters (k={k})',
        opacity=0.7,
        height=700,
        labels={x_col: x_col, y_col: y_col, z_col: z_col}
    )
    
    fig.update_layout(scene=dict(xaxis_title=x_col, yaxis_title=y_col, zaxis_title=z_col))
    fig.show()
    
    return data
```
