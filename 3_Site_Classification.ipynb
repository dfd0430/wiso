```python
import pandas as pd

# Load data from Part 1
charging_sessions = pd.read_parquet('intermediate_data/charging_sessions.parquet')
```

### 3. Site Classification

In this section we use 4 parameters to figure out which site is which them being:
- amount of charging sessions without account
- average duration of session
- standard deviation of duration
- percentage of session on the weekend

We conclude that Site 1 is the private workplace because it has far fewer unregistered users, indicating a closed system where most drivers hold permanent accounts. The weekend usage is also negligible at only 2%, which is typical for a corporate office. Finally, the longer average session times and lower standard deviation suggest a stable, predictable environment that aligns with a standard 9-to-5 workday.

```python
charging_sessions_site1=charging_sessions[charging_sessions['site_id']==1].copy()
charging_sessions_site2=charging_sessions[charging_sessions['site_id']==2].copy()

empty_user_count_1=(charging_sessions_site1["user_id"].isna()).sum()
empty_user_count_2=(charging_sessions_site2["user_id"].isna()).sum()
print(f"amount of charging sessions without account at site 1: {empty_user_count_1}")
print(f"amount of charging sessions without account at site 2: {empty_user_count_2}\n")


charging_sessions_site1["duration"] = charging_sessions_site1["disconnect_time"] - charging_sessions_site1["connection_time"]
charging_sessions_site2["duration"] = charging_sessions_site2["disconnect_time"] - charging_sessions_site2["connection_time"]


print(f"average duration at site 1: {charging_sessions_site1["duration"].mean()}")
print(f"average duration at site 2: {charging_sessions_site2["duration"].mean()}\n")



std_site1 = charging_sessions_site1["duration"].dt.total_seconds().std()
std_site2 = charging_sessions_site2["duration"].dt.total_seconds().std()

print(f"Site 1 standard deviation: +/- {std_site1 / 60 :.0f} minutes")
print(f"Site 2 standard deviation: +/- {std_site2 / 60 :.0f} minutes\n")


weekends_site1 = charging_sessions_site1[charging_sessions_site1['connection_time'].dt.dayofweek >= 5].copy()
weekends_site2 = charging_sessions_site2[charging_sessions_site2['connection_time'].dt.dayofweek >= 5].copy()

weekends_site1_percent_of_sessions = weekends_site1["id"].shape[0] / charging_sessions_site1.shape[0]
weekends_site2_percent_of_sessions = weekends_site2["id"].shape[0] / charging_sessions_site2.shape[0]

print(f"percentage of charging sessions on weekends for site 1: {weekends_site1_percent_of_sessions}")
print(f"percentage of charging sessions on weekends for site 2: {weekends_site2_percent_of_sessions}")
```
